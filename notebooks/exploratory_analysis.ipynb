# ChurnSense AI — Exploratory Data Analysis
This notebook explores the included **synthetic** customer churn dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('../data/customer_churn.csv')
df.head()
```

```python
df.info()
df.isna().sum()
```

```python
df['churn'].value_counts(normalize=True).rename('share')
```

```python
sns.countplot(data=df, x='contract_type', hue='churn')
plt.title('Synthetic churn by contract type')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## Interpretation
Patterns in this generated dataset are illustrative only. They should not be generalized to real customers.
